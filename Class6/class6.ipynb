```python
# UNIFORM DISTRIBUTION
# np.random.rand() generates a random number between 0 and 1. It can also generate an array of random numbers if you specify the shape.
# np.random.seed() is used to set the seed for the random number generator. This ensures that the random numbers generated are reproducible. If you set the same seed, you will get the same random numbers every time you run the code.
# NORMAL DISTRIBUTION
# np.random.randn() generates random numbers from a standard normal distribution (mean = 0, standard deviation = 1). It can also generate an array of random numbers if you specify the shape.
print(np.random.seed(5))
print(np.random.rand(10))
print(np.random.seed(5))
print(np.random.randn(10))
```

```python
# np.random.randint() generates random integers between a specified range. It can also generate an array of random integers if you specify the shape.
# np.random.randint(low=0, high=10, size=(5,3))
print(np.random.randint(0,10,(5,3)))
```

```python
# np.random.choice() generates a random sample from a given 1-D array. It can also generate an array of random samples if you specify the shape.
# np.random.shuffle() randomly shuffles the elements of a given array in place. It does not return a new array, but modifies the original array.
lst=[1,2,3,4,5]
# np.random.choice(lst, size=3, replace=False,probability) # replace=False means that the same element cannot be selected more than once
print(np.random.choice(lst,3,replace=True, p=[0.1, 0.2, 0.3, 0.2, 0.2])) # replace=True means that the same element can be selected more than once
print(np.random.choice(lst,3,replace=False)) # replace=False means that the same element cannot be selected more than once
np.random.shuffle(lst)
print(lst)
```

```python
# min max mean std median sum are some of the statistical functions that can be performed on an array. These functions can be used to get a better understanding of the data and its distribution.
x=np.arange(35,100).reshape(5,13)
print(x)
print(np.mean(x))
print(np.max(x))
print(x.min(axis=0))
print(np.std(x))
print(np.median(x))
print(np.sum(x))
# for perticular row and columb
print(np.mean(x,axis=1)) # axis=0 means that the function will be applied to each column
```
